```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
```

```python
data='youtube_ad_revenue_dataset.csv'
df=pd.read_csv(data)
df.head()
```

```python
df.info()
```

```python

df['time']=pd.to_datetime(df['date']).dt.time


df['date']=pd.to_datetime(df['date']).dt.date

```

```python
df.head()
```

```python
100*df.isnull().sum()/len(df)
```

```python
df['likes']=df['likes'].fillna(df['likes'].median())
```

```python
df['comments']=df['comments'].fillna(df['comments'].median())
df['watch_time_minutes']=df['watch_time_minutes'].fillna(df['watch_time_minutes'].median())
```

```python
100*df.isnull().sum()/len(df)
df.head()
```

```python
df1 = df.drop('video_id', axis=1)
```

```python
df1.head()
```

```python
df1.boxplot(column='views')
plt.show()

df1.boxplot(column='likes')
plt.show()

df1.boxplot(column='comments')
plt.show()

df1.boxplot(column='subscribers')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df1.corr(numeric_only=True),annot=True,cmap='coolwarm',fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder
print("Scikit-learn installed successfully")
```

```python
le_category = LabelEncoder()
le_device = LabelEncoder()
le_country = LabelEncoder()

df1['category'] = le_category.fit_transform(df1['category'])
df1['device'] = le_device.fit_transform(df1['device'])
df1['country'] = le_country.fit_transform(df1['country'])
```

```python
df1.info()
```

```python
x=df1.drop(['ad_revenue_usd','date','time'],axis=1)
y=df1['ad_revenue_usd']
```

```python
x.select_dtypes(include='object').columns
```

```python
x.info()
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(x_train_scaled, y_train)
```

```python
y_pred = lr.predict(x_test_scaled)
```

### Linear Regression

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
```

### Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor

dt = DecisionTreeRegressor(random_state=42)

dt.fit(x_train, y_train)

y_pred_dt = dt.predict(x_test)
```

```python
print("R2:", r2_score(y_test, y_pred_dt))
print("MAE:", mean_absolute_error(y_test, y_pred_dt))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_dt)))
```

### Random Forest

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf.fit(x_train, y_train)

y_pred_rf = rf.predict(x_test)
```

```python
print("R2:", r2_score(y_test, y_pred_rf))
print("MAE:", mean_absolute_error(y_test, y_pred_rf))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_rf)))
```

### Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingRegressor

gb = GradientBoostingRegressor(random_state=42)

gb.fit(x_train, y_train)

y_pred_gb = gb.predict(x_test)

print("R2:", r2_score(y_test, y_pred_gb))
print("MAE:", mean_absolute_error(y_test, y_pred_gb))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_gb)))
```
